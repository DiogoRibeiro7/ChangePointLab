# Multi-method changepoint comparison

This notebook demonstrates how different detectors in the changepoint toolkit behave on a variety of synthetic datasets. Each scenario highlights different kinds of changepoints (abrupt, gradual, variance, periodic, compositional) and shows how BOCPD, PELT, E-Divisive, HSMM/SD-HMM, and Within-Period CPD can be applied. Performance metrics and plots help illustrate which method works best for a given situation.

```
from examples.multi_method_comparison import (
    abrupt_mean_shift,
    variance_shift,
    gradual_mean_change,
    periodic_binary,
    compositional_example,
    run_bocpd,
    run_bocpd_boosted,
    run_pelt_known,
    run_pelt_unknown,
    run_edivisive_default,
    run_edivisive_strict,
    run_hsmm,
    run_sdhmm,
    run_within_period,
    _plot_series,
)
import matplotlib.pyplot as plt
```

## Abrupt mean change

```
data, binary, true_cps = abrupt_mean_shift()
fig, ax = plt.subplots(3, 2, figsize=(12,8))
ax = ax.ravel()
_plot_series(ax[0], binary, run_bocpd(binary), 'BOCPD')
_plot_series(ax[1], binary, run_bocpd_boosted(binary), 'BOCPD boosted')
_plot_series(ax[2], data, run_pelt_known(data), 'PELT known var')
_plot_series(ax[3], data, run_pelt_unknown(data), 'PELT unknown var')
_plot_series(ax[4], data, run_edivisive_default(data), 'E-Divisive R10')
_plot_series(ax[5], data, run_edivisive_strict(data), 'E-Divisive R100')
fig.suptitle(f'True CPs {true_cps}')
plt.tight_layout()
```

HSMM estimate:

```
print(run_hsmm(data))
```

## Periodic binary data

```
data, binary, true_cps = periodic_binary()
fig, ax = plt.subplots(3, 2, figsize=(12,8))
ax = ax.ravel()
_plot_series(ax[0], binary, run_bocpd(binary), 'BOCPD')
_plot_series(ax[1], binary, run_bocpd_boosted(binary), 'BOCPD boosted')
_plot_series(ax[2], data, run_pelt_known(data), 'PELT known var')
_plot_series(ax[3], data, run_pelt_unknown(data), 'PELT unknown var')
_plot_series(ax[4], data, run_edivisive_default(data), 'E-Divisive R10')
_plot_series(ax[5], data, run_edivisive_strict(data), 'E-Divisive R100')
fig.suptitle(f'True CPs {true_cps}')
plt.tight_layout()
```

Within-period CPD estimate:

```
print(run_within_period(binary, period=24))
```

### Discussion

- **BOCPD** is well suited for online detection of binary events and reacts
  quickly to abrupt probability shifts.
- **PELT** provides optimal offline segmentation for Gaussian data and works
  well when change points are sharp.
- **E-Divisive** handles nonparametric distributions and remains effective
  under variance shifts.
- **HSMM/HMM** model latent states and durations, capturing gradual changes
  through state transitions.
- **SD-HMM** specializes in compositional observations by enforcing simplex
  constraints.
- **Within-period** detection leverages known periodicity to discover phase
  specific changes on a circular time axis.
